```python
import pandas as pd

df2020 = pd.read_csv(r"C:\workspace\displacement_index\displacement_index_current\09-Proximity-to-Current-or-Future-HCT\Output\hct_area_coverage_2023.csv", dtype={"GEOID": str})
crosswalk = pd.read_csv(r"c:\workspace\displacement_index\displacement_index_current\nhgis_tr2010_tr2020_53.csv",
                        dtype={"GEOID_2010": str, "GEOID_2020": str})
# crosswalk.groupby("tr2010ge")["wt_pop"].sum().describe()
df2020.columns
# crosswalk.columns
```
